```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
data = {
    'GRE': [380, 660, 800, 640, 520, 760, 560, 400, 700, 480],
    'GPA': [3.0, 3.4, 4.0, 3.5, 3.2, 3.8, 2.9, 2.5, 3.9, 3.1],
    'Rank': [3, 1, 2, 2, 3, 1, 4, 4, 1, 3],
    'Admitted': [0, 1, 1, 1, 0, 1, 0, 0, 1, 0]
}
df = pd.DataFrame(data)
X = df[['GRE', 'GPA', 'Rank']]
y = df['Admitted']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
new_student = pd.DataFrame({'GRE': [700], 'GPA': [3.6], 'Rank': [2]})
prediction = model.predict(new_student)
print(f"Prediction for new student: {'Admitted' if prediction[0] == 1 else 'Not Admitted'}")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
np.random.seed(42)
num_transactions = 500
df = pd.DataFrame({
    'amount': np.random.rand(num_transactions) * 500,
    'features_1': np.random.rand(num_transactions) * 10,
    'is_fraud': np.random.randint(0, 2, num_transactions) # Simplified fraud (50/50 for brevity)
})
df.loc[df['is_fraud'] == 1, 'amount'] += 200
X = df.drop('is_fraud', axis=1)
y = df['is_fraud']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred))
new_transaction = pd.DataFrame([[300, 8]], columns=X.columns)
prediction = model.predict(new_transaction)
probability = model.predict_proba(new_transaction)[:, 1]
print(f"\nNew transaction: {new_transaction.iloc[0].to_dict()}")
print(f"Predicted Fraud: {prediction[0]} (Probability: {probability[0]:.2f})")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
np.random.seed(42)
num_applicants = 100
data = {
    'credit_score': np.random.randint(300, 850, num_applicants),
    'income': np.random.randint(20000, 100000, num_applicants),
    'loan_amount': np.random.randint(50000, 500000, num_applicants),
    'loan_status': np.random.randint(0, 2, num_applicants) # 0=Rejected, 1=Approved
}
df = pd.DataFrame(data)
df.loc[df['loan_status'] == 1, 'credit_score'] += 100
df.loc[df['loan_status'] == 0, 'income'] -= 10000
print("--- Dummy Data Head ---")
print(df.head())
print(f"\nLoan Status counts:\n{df['loan_status'].value_counts()}")
X = df.drop('loan_status', axis=1)
y = df['loan_status']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("\n--- Training Logistic Regression Model ---")
model = LogisticRegression(random_state=42, solver='liblinear') # 'liblinear' is good for small datasets
model.fit(X_train, y_train)
print("\n--- Evaluation ---")
y_pred = model.predict(X_test)
print("Classification Report:")
print(classification_report(y_test, y_pred))
print("\n--- Predicting for a New Applicant ---")
new_applicant = pd.DataFrame([[750, 80000, 200000]],
                             columns=['credit_score', 'income', 'loan_amount'])
prediction = model.predict(new_applicant)
probability = model.predict_proba(new_applicant)[:, 1] # Probability of approval (class 1)
print(f"\nNew applicant details: {new_applicant.iloc[0].to_dict()}")
print(f"Predicted Loan Status (0=Rejected, 1=Approved): {prediction[0]}")
print(f"Probability of Approval: {probability[0]:.2f}")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
data = {
    'Tenure': [1, 34, 2, 45, 5, 22, 52, 10, 3, 60],
    'MonthlyCharges': [29.85, 56.95, 53.85, 42.30, 70.70, 99.65, 89.10, 29.75, 104.80, 56.15],
    'TotalCharges': [29.85, 1889.50, 108.15, 1840.75, 353.30, 2090.45, 4580.20, 301.90, 304.20, 3400.10],
    'Churn': [1, 0, 1, 0, 1, 0, 0, 1, 1, 0]
}
df = pd.DataFrame(data)
X = df[['Tenure', 'MonthlyCharges', 'TotalCharges']]
y = df['Churn']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
model = LogisticRegression()
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
new_customer = pd.DataFrame({'Tenure': [12], 'MonthlyCharges': [75.0], 'TotalCharges': [900.0]})
new_customer_scaled = scaler.transform(new_customer)
churn_prediction = model.predict(new_customer_scaled)
print(f"Prediction for new customer: {'Churn' if churn_prediction[0] == 1 else 'No Churn'}")
```
